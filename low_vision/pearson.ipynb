# Correlação de Pearson

### Bibliotecas utilizadas

```python
import pandas as pd
import numpy as np
from scipy import stats
```

### Lendo a base

```python
caminho = r'C:\Users\saulosgil\OneDrive\Documentos\functional_tests_blinding\data\data_blinding.xlsx'
df = pd.read_excel(caminho)

df.head()
```

## **Selecionar apenas os baixa visão**

```python
df = df[df['baixa_visao'] == 1]
df.head()
```

## **Time Up and Go**

```python
# Exemplo de dados (substitua pelos seus)
# x = np.array([10, 20, 30, 40, 50])
# y = np.array([12, 24, 33, 47, 55])
x = np.array(df['tug_1'])
y = np.array(df['tug_2'])

# --- Validação de entrada ---
if x.size != y.size:
    raise ValueError("As listas/arrays devem ter o mesmo tamanho.")
if x.size < 2:
    raise ValueError("É necessário pelo menos 2 pontos para calcular correlação.")

# --- Correlação de Pearson ---
pearson_r, pearson_p = stats.pearsonr(x, y)

# --- Correlação de Spearman ---
spearman_r, spearman_p = stats.spearmanr(x, y)

# --- Resultados ---
print("Correlação de Pearson:")
print(f"  r = {pearson_r:.4f}, p-valor = {pearson_p:.4e}")

print("\nCorrelação de Spearman:")
print(f"  r = {spearman_r:.4f}, p-valor = {spearman_p:.4e}")

# --- Interpretação simples ---
def interpreta_r(r):
    r_abs = abs(r)
    if r_abs < 0.3:
        return "Correlação fraca"
    elif r_abs < 0.7:
        return "Correlação moderada"
    else:
        return "Correlação forte"

print("\nInterpretação Pearson:", interpreta_r(pearson_r))
print("Interpretação Spearman:", interpreta_r(spearman_r))
```
